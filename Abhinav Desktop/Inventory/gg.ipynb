```python
from docplex.mp.model import Model
import numpy as np
```

```python
mdl = Model('LR')


x1 = mdl.continuous_var(lb=0, ub=1, name='x1')
x2 = mdl.continuous_var(lb=0, ub=1, name='x2')
x3 = mdl.continuous_var(lb=0, ub=1, name='x3')
x4 = mdl.continuous_var(lb=0, ub=1, name='x4')

# Objective: minimize travel distance
mdl.maximize(mdl.sum(16*x1+ 10*x2+4*x4))

# Each city has exactly one outgoing arc

mdl.add_constraint((8*x1+2*x2+x3+4*x4) <= 10)
mdl.add_constraint((x1+x2) <= 1)
mdl.add_constraint((x3+x4) <= 1)

solution = mdl.solve(log_output=True)

```

```python
mdl = Model('LR')


x1 = mdl.continuous_var(lb=0, ub=1, name='x1')
x2 = mdl.continuous_var(lb=0, ub=1, name='x2')
x3 = mdl.continuous_var(lb=0, ub=1, name='x3')
x4 = mdl.continuous_var(lb=0, ub=1, name='x4')
u = 10000

# Objective: minimize travel distance
mdl.maximize(mdl.sum(16*x1+ 10*x2+4*x4-u*(8*x1+2*x2+x3+4*x4 - 10)))

# Each city has exactly one outgoing arc

# mdl.add_constraint((8*x1+2*x2+x3+4*x4) <= 10)
mdl.add_constraint((x1+x2) <= 1)
mdl.add_constraint((x3+x4) <= 1)

solution = mdl.solve(log_output=True)

```

```python
# Redefine the constrained model (linear relaxed problem)
mdl_constrained = Model('LR_constrained')

x1_c = mdl_constrained.continuous_var(lb=0, ub=1, name='x1')
x2_c = mdl_constrained.continuous_var(lb=0, ub=1, name='x2')
x3_c = mdl_constrained.continuous_var(lb=0, ub=1, name='x3')
x4_c = mdl_constrained.continuous_var(lb=0, ub=1, name='x4')

# Objective: maximize
mdl_constrained.maximize(16*x1_c + 10*x2_c + 4*x4_c)

# Constraints
mdl_constrained.add_constraint((8*x1_c + 2*x2_c + x3_c + 4*x4_c) <= 10)
mdl_constrained.add_constraint((x1_c + x2_c) <= 1)
mdl_constrained.add_constraint((x3_c + x4_c) <= 1)

solution_constrained = mdl_constrained.solve(log_output=True)

# Compare solutions
print("Constrained solution objective:", solution_constrained.objective_value)
print("Penalized solution objective:", solution.objective_value)
print("Constrained x1:", solution_constrained[x1_c])
print("Penalized x1:", solution[x1])
print("Constrained x2:", solution_constrained[x2_c])
print("Penalized x2:", solution[x2])
print("Constrained x3:", solution_constrained[x3_c])
print("Penalized x3:", solution[x3])
print("Constrained x4:", solution_constrained[x4_c])
print("Penalized x4:", solution[x4])

# Check if penalized solution satisfies the removed constraint
constraint_value = 8*solution[x1] + 2*solution[x2] + solution[x3] + 4*solution[x4]
print("Penalized constraint value (should be <=10):", constraint_value)
```
