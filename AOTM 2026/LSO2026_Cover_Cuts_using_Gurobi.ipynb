## 2. Imports and problem data

```python
import gurobipy as gp
from gurobipy import GRB
import pandas as pd
import matplotlib.pyplot as plt

profit = {
    1: 13,
    2: 8,
    3: 7,
    4: 7,
    5: 6,
    6: 5,
    7: 1,
}

weight = {
    1: 11,
    2: 6,
    3: 6,
    4: 5,
    5: 5,
    6: 4,
    7: 1,
}

capacity = 19
J = list(range(1, 8))
```

```python
m = gp.Model("knapsack")
m.setParam("Cuts", 0)


#Define Variables
x = {
        j: m.addVar(lb=0.0, ub=5.0, vtype=GRB.CONTINUOUS, name=f"x{j}")
        #vtype=GRB.BINARY, GRB.CONTINUOUS
        for j in J
    }

m.addConstr(
        gp.quicksum(weight[j] * x[j] for j in J) >= capacity,





    )
#Add and experiment with cover cuts
#---------------
#---------------

m.setObjective(
      gp.quicksum(profit[j] * x[j] for j in J),
      GRB.MINIMIZE
      #GRB.MAXIMIZE, GRB.MINIMIZE

  )

m.optimize()
sol = {j: x[j].X for j in J}


```

```python
print(f"Objective: {m.ObjVal:.6f}")
print("Decision Variables:")
for j, value in sol.items():
    print(f"x{j} = {value:.3f}")
```

```python
#Input to seperation problem
print("LP optimal solution:")
y_bar = {}
for j, value in sol.items():
    y_bar[j] = value
x_bar = list(y_bar.values())
print(x_bar)

# Perform element-wise subtraction
ones_list = [1] * 7
obj_coef = [ones_list[i] - x_bar[i] for i in range(len(ones_list))]
print(obj_coef)
```

```python
#Code your Seperation problem here


# 1. Create a new optimization model
s = gp.Model("Seperation")

# 2. define decision variables y

y = {
        j: s.addVar(lb=0.0, ub=1.0, vtype= ----, name=f"y{j}")
        #vtype=GRB.BINARY, GRB.CONTINUOUS
        for j in J
    }


# 3. Add the constraint

#----------------

# 4. Define Objective and Set the objective function to minimize


s.setObjective(
      gp.quicksum( obj_coef[j-1]* y[j] for j in J),
      GRB.MINIMIZE
  )



# 5. Solve the model using Gurobi's optimization engine
s.optimize()


```

```python
# 6. Print the optimal values of the variables and objective function



print(f"Objective: {s.ObjVal:.6f}")
print("Decision Variables:")
# Iterate through the decision variables of the 's' model
for j in J:
    print(f"y{j} = {y[j].X:.3f}")
```
